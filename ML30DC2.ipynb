# Creating a Series and DataFrame

```python
import pandas as pd
import numpy as np

# Creating a Series by passing a list of values
s = pd.Series([1, 3, 5, np.nan, 6, 8])
print("Series:")
print(s)
```

```python
# Creating a DataFrame by passing a NumPy array, with a datetime index and labeled columns
dates = pd.date_range("20130101", periods=6)
df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list("ABCD"))
print("\nDataFrame:")
print(df)
```

# Viewing Data

```python
print("First 5 rows:")
print(df.head())
```

```python
print("\nLast 3 rows:")
print(df.tail(3))
```

```python
print("\nIndex:")
print(df.index)
```

```python
print("\nColumns:")
print(df.columns)
```

```python
print("\nNumPy data:")
print(df.to_numpy())
```

# Selection

```python
# Selecting a single column
print("Column 'A':")
print(df["A"])
```

```python
# Selecting by slicing rows
print("\nRows from index 0 to 2:")
print(df[0:3])
```

```python
# Selecting by label
print("\nRow with index '20130102':")
print(df.loc[dates[1]])
```

```python
# Selecting by position
print("\nValue at row 0, column 0:")
print(df.iloc[0, 0])
```

# Missing Data

```python
df1 = df.copy()
df1.iloc[0, 0] = np.nan
print("DataFrame with a missing value:")
print(df1)
```

```python
# Dropping rows with missing data
print("\nDataFrame after dropping rows with missing data:")
print(df1.dropna(how="any"))
```

```python
# Filling missing data
print("\nDataFrame after filling missing data with 0:")
print(df1.fillna(value=0))
```

# Operations

```python
# Descriptive statistics
print("Descriptive statistics:")
print(df.describe())
```

```python
# Applying a function to the data
print("\nApplying a cumulative sum function:")
print(df.apply(np.cumsum))
```

# Grouping

```python
df = pd.DataFrame(
    {
        "A": ["foo", "bar", "foo", "bar", "foo", "bar", "foo", "foo"],
        "B": ["one", "one", "two", "three", "two", "two", "one", "three"],
        "C": np.random.randn(8),
        "D": np.random.randn(8),
    }
)

print("Original DataFrame for grouping:")
print(df)
```

```python
print("\nGrouping by column 'A' and applying sum:")
print(df.groupby("A").sum())
```

```python
print("\nGrouping by multiple columns ('A' and 'B') and applying sum:")
print(df.groupby(["A", "B"]).sum())
```

# Reshaping

## Stack

The `.stack()` method "compresses" a level in the DataFrame's columns to produce a Series (or reduce the level of a MultiIndex).

```python
tuples = list(
    zip(
        ["bar", "bar", "baz", "baz", "foo", "foo", "qux", "qux"],
        ["one", "two", "one", "two", "one", "two", "one", "two"],
    )
)

index = pd.MultiIndex.from_tuples(tuples, names=["first", "second"])
df_multiindex = pd.DataFrame(np.random.randn(8, 2), index=index, columns=["A", "B"])
df_multiindex2 = df_multiindex[:4]

print("Original DataFrame with MultiIndex:")
print(df_multiindex2)
```

```python
print("\nStacked DataFrame:")
stacked = df_multiindex2.stack()
print(stacked)
```

## Unstack

The `.unstack()` method does the opposite of `.stack()`, "uncompressing" a level in the index to form the columns.

```python
print("Unstacked Series:")
print(stacked.unstack())
```

```python
print("\nUnstacked Series (unstacking the second level):")
print(stacked.unstack(1))
```

```python
print("\nUnstacked Series (unstacking the first level):")
print(stacked.unstack(0))
```
