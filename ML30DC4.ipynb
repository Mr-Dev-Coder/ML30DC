# Data loading


```python
import seaborn as sns

df = sns.load_dataset('tips')
display(df.head())
```

# Distribution plots

```python
sns.histplot(data=df, x='total_bill')
```

```python
sns.kdeplot(data=df, x='total_bill')
```

```python
sns.displot(data=df, x='total_bill', kde=True)
```

# Categorical plots

```python
import matplotlib.pyplot as plt

# Create a box plot of 'total_bill' by 'day'
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='day', y='total_bill')
plt.title('Box Plot of Total Bill by Day')
plt.show()
```

```python
# Create a violin plot of 'total_bill' by 'day'
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='day', y='total_bill')
plt.title('Violin Plot of Total Bill by Day')
plt.show()
```

```python
# Create a swarm plot of 'total_bill' by 'day'
plt.figure(figsize=(8, 5))
sns.swarmplot(data=df, x='day', y='total_bill')
plt.title('Swarm Plot of Total Bill by Day')
plt.show()
```

```python
# Create a count plot of 'day'
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='day')
plt.title('Count Plot of Day')
plt.show()
```

# Regression plots

```python
import matplotlib.pyplot as plt

# Create a scatter plot with a regression line
plt.figure(figsize=(8, 6))
sns.regplot(data=df, x='total_bill', y='tip')
plt.title('Regression Plot of Total Bill vs. Tip')
plt.show()
```

```python
# Create a scatter plot with separate regression lines for each 'smoker' category
sns.lmplot(data=df, x='total_bill', y='tip', col='smoker', height=5, aspect=0.8)
plt.suptitle('Regression Plot of Total Bill vs. Tip by Smoker Status', y=1.02)
plt.show()
```

```python
# Create a scatter plot with regression lines for each 'day'
sns.lmplot(data=df, x='total_bill', y='tip', col='day', height=5, aspect=0.6)
plt.suptitle('Regression Plot of Total Bill vs. Tip by Day', y=1.02)
plt.show()
```

# Matrix plots

```python
correlation_matrix = df.corr(numeric_only=True)
print("Correlation Matrix:")
display(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.clustermap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", figsize=(6, 6))
plt.suptitle('Correlation Clustermap', y=1.02)
plt.show()
```

# Grids

```python
import matplotlib.pyplot as plt

# Create a FacetGrid for the first plot
g1 = sns.FacetGrid(df, col='time')
g1.map(sns.scatterplot, 'total_bill', 'tip')
g1.fig.suptitle('Scatter Plot of Total Bill vs. Tip by Time of Day', y=1.02)
plt.show()
```

```python
# Create another FacetGrid for the second plot
g2 = sns.FacetGrid(df, col='day', row='smoker')
g2.map(sns.histplot, 'total_bill')
g2.fig.suptitle('Distribution of Total Bill by Day and Smoker Status', y=1.02)
plt.show()
```

# Styling

```python
import matplotlib.pyplot as plt

# 1. Set a different Seaborn style and plot it
sns.set_style('darkgrid')
print("Style set to 'darkgrid'")

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='total_bill', y='tip')
plt.title('Scatter Plot with Darkgrid Style')
plt.show()
```

```python
# 2. Use seaborn.set_palette() to change the color palette and plot it
sns.set_palette('viridis')
print("Color palette set to 'viridis'")

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='day')
plt.title('Scatter Plot with Viridis Palette (colored by Day)')
plt.show()
```

```python
# 3. Reset the Seaborn style and palette to the default settings  and plot it with default
sns.set_style('whitegrid')
sns.set_palette('deep')
print("Style and palette reset to defaults")

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='total_bill', y='tip')
plt.title('Scatter Plot with Default Styling')
plt.show()
```
